```python
Traffic = datasets["traffic"]
Traffic
```

## Fixing Data Types

```python
Traffic['timestamp'] = pd.to_datetime(Traffic['timestamp'], format='%Y-%m-%d %H:%M:%S')
```

<a id="handling-missing-data"></a>
## Handling Missing Values

---
[⬆️ Back to Top](#table-of-contents)

```python
Traffic.isna().sum()
```

```python
Traffic[Traffic.isna().any(axis=1)]

Traffic2 = Traffic.dropna()
print(len(Traffic) - len(Traffic2))
Traffic[Traffic.isna().any(axis=1)]
```

##### the missing values are all on the same rows, so we have to drop them, now no missing values

```python
Traffic = Traffic.dropna()
Traffic.isnull().sum()
```

<a id="handling-duplicates"></a>
## Handling Duplicates

```python
print(Traffic.duplicated().sum())
Traffic.drop_duplicates(inplace=True)
```

## Handling Outliers

```python
Traffic.describe()
```

```python
numerical_columns = Traffic.select_dtypes(include=[np.number]).columns.tolist()

for col in numerical_columns:
    q1 = Traffic[col].quantile(0.25)
    q3 = Traffic[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = Traffic[(Traffic[col] < lower_bound) | (Traffic[col] > upper_bound)]
    outliers_percentage = len(outliers) / len(Traffic) * 100
    print(f"Number of outliers in {col}: {len(outliers)}")
    print(f"Percentage of outliers in {col}: {outliers_percentage:.2f}% \n")
```

```python
q1 = Traffic['traffic_volume'].quantile(0.25)
q3 = Traffic['traffic_volume'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
outliers = Traffic[(Traffic['traffic_volume'] < lower_bound) | (Traffic['traffic_volume'] > upper_bound)]
outliers_percentage = len(outliers) / len(Traffic) * 100
print(f"Number of outliers in traffic_volume: {len(outliers)}")
print(f"Percentage of outliers in traffic_volume: {outliers_percentage:.2f}% \n")
# Traffic['traffic_volume'] = Traffic['traffic_volume'].clip(lower=lower_bound, upper=upper_bound)
outliers
```

### we'll keep it as it is for now as the sensors are working properly, so this is a correct measurement
### but we have to be careful when modeling on this data!!!

## checking data consistency

```python
for col in Traffic.columns:
    if Traffic[col].dtype == 'str':
        print(f"Column: {col}")
        print(Traffic[col].unique())
        print("\n")
```

### the data is consistent, we can proceed

# ⏰ 2. When does traffic congestion reach its highest levels, and what temporal patterns characterize these peaks?

```python
# Ensure temporal feature extraction on cleaned traffic data
Traffic['hour'] = Traffic['timestamp'].dt.hour
Traffic['day_of_week'] = Traffic['timestamp'].dt.day_name()
Traffic['day_num'] = Traffic['timestamp'].dt.dayofweek

dow_order = [
    'Monday',
    'Tuesday',
    'Wednesday',
    'Thursday',
    'Friday',
    'Saturday',
    'Sunday',
]


def assign_period(h):
  if 7 <= h <= 9:
    return 'Morning Peak (07-09)'
  elif 16 <= h <= 19:
    return 'Evening Peak (16-19)'
  elif 10 <= h <= 15:
    return 'Mid-day Off-Peak (10-15)'
  elif 20 <= h <= 22:
    return 'Late Evening (20-22)'
  else:
    return 'Night / Early Morning (23-06)'


Traffic['time_period'] = Traffic['hour'].apply(assign_period)

# Plot Setup
plt.style.use('seaborn-v0_8-whitegrid')
fig, axes = plt.subplots(3, 2, figsize=(16, 18))
plt.subplots_adjust(hspace=0.35, wspace=0.25)

# 1. Hourly Line Plot
hourly_df = (
    Traffic.groupby('hour')['congestion_index'].mean().reset_index()
)
sns.lineplot(
    data=hourly_df,
    x='hour',
    y='congestion_index',
    marker='o',
    color='#2563eb',
    linewidth=2.5,
    ax=axes[0, 0],
)
axes[0, 0].set_title(
    'Average Congestion Index by Hour of Day', fontsize=14, fontweight='bold'
)
axes[0, 0].set_xlabel('Hour of Day', fontsize=12)
axes[0, 0].set_ylabel('Mean Congestion Index', fontsize=12)
axes[0, 0].set_xticks(range(0, 24))

# 2. Weekday vs Weekend Comparison
wknd_df = (
    Traffic.groupby(['hour', 'weekend_flag'])['congestion_index']
    .mean()
    .reset_index()
)
wknd_df['Day Type'] = wknd_df['weekend_flag'].map({0.0: 'Weekday', 1.0: 'Weekend'})
sns.lineplot(
    data=wknd_df,
    x='hour',
    y='congestion_index',
    hue='Day Type',
    marker='o',
    palette=['#2563eb', '#f59e0b'],
    linewidth=2.5,
    ax=axes[0, 1],
)
axes[0, 1].set_title(
    'Hourly Congestion: Weekday vs. Weekend', fontsize=14, fontweight='bold'
)
axes[0, 1].set_xlabel('Hour of Day', fontsize=12)
axes[0, 1].set_ylabel('Mean Congestion Index', fontsize=12)
axes[0, 1].set_xticks(range(0, 24))

# 3. Day of Week Bar Plot
dow_df = (
    Traffic.groupby('day_of_week')['congestion_index']
    .mean()
    .reindex(dow_order)
    .reset_index()
)
sns.barplot(
    data=dow_df, x='day_of_week', y='congestion_index', palette='Blues_r', ax=axes[1, 0]
)
axes[1, 0].set_title(
    'Average Congestion Index by Day of Week', fontsize=14, fontweight='bold'
)
axes[1, 0].set_xlabel('Day of Week', fontsize=12)
axes[1, 0].set_ylabel('Mean Congestion Index', fontsize=12)
axes[1, 0].tick_params(axis='x', rotation=30)

# 4. Rush-Hour Comparison
rush_df = (
    Traffic.groupby('rush_hour_flag')['congestion_index'].mean().reset_index()
)
rush_df['Rush Hour'] = rush_df['rush_hour_flag'].map(
    {0.0: 'Non-Rush Hour', 1.0: 'Rush Hour'}
)
sns.barplot(
    data=rush_df,
    x='Rush Hour',
    y='congestion_index',
    palette=['#64748b', '#ef4444'],
    ax=axes[1, 1],
)
axes[1, 1].set_title(
    'Congestion Index: Rush Hour vs. Non-Rush Hour',
    fontsize=14,
    fontweight='bold',
)
axes[1, 1].set_xlabel('', fontsize=12)
axes[1, 1].set_ylabel('Mean Congestion Index', fontsize=12)

# 5. Heatmap Hour x Day
days_map = {0: 'Mon', 1: 'Tue', 2: 'Wed', 3: 'Thu', 4: 'Fri', 5: 'Sat', 6: 'Sun'}
heatmap_matrix = (
    Traffic.groupby(['day_num', 'hour'])['congestion_index'].mean().unstack()
)
heatmap_matrix.index = [days_map[i] for i in heatmap_matrix.index]
sns.heatmap(
    heatmap_matrix,
    cmap='YlOrRd',
    cbar_kws={'label': 'Congestion Index'},
    ax=axes[2, 0],
)
axes[2, 0].set_title(
    'Heatmap: Congestion by Hour × Day of Week', fontsize=14, fontweight='bold'
)
axes[2, 0].set_xlabel('Hour of Day', fontsize=12)
axes[2, 0].set_ylabel('Day of Week', fontsize=12)

# 6. Peak-Period Analysis
period_order = [
    'Morning Peak (07-09)',
    'Evening Peak (16-19)',
    'Mid-day Off-Peak (10-15)',
    'Late Evening (20-22)',
    'Night / Early Morning (23-06)',
]
period_df = (
    Traffic.groupby('time_period')['congestion_index']
    .mean()
    .reindex(period_order)
    .reset_index()
)
sns.barplot(
    data=period_df,
    y='time_period',
    x='congestion_index',
    palette='Reds_r',
    ax=axes[2, 1],
)
axes[2, 1].set_title(
    'Congestion Index Across Temporal Periods', fontsize=14, fontweight='bold'
)
axes[2, 1].set_xlabel('Mean Congestion Index', fontsize=12)
axes[2, 1].set_ylabel('', fontsize=12)

plt.tight_layout()
plt.show()
```

### 📊 Key Insights: Traffic Congestion Patterns

* **Bi-Modal Daily Peaks:** Congestion strongly concentrates during morning (**07:00–09:00**) and evening (**16:00–19:00**) rush hours.
* **Weekday vs. Weekend Dichotomy:** Weekdays bear the brunt of traffic demand, whereas weekend periods show a sharp drop in congestion intensity.
* **Peak-Volume Compound Effect:** The absolute highest congestion indices occur specifically at the intersection of **weekdays and peak commute windows**, confirming work/school travel as the primary systemic bottleneck.
---

### 💡 Simple Solutions to Fix Peak Traffic

1. **Flexible Work & School Hours:** Encourage workplaces and schools to offer flexible start times or remote work options to spread out the rush hour crowd.
2. **Smarter Traffic Lights:** Install smart sensors on traffic lights so green lights stay on longer where cars are piling up most.
3. **More Buses During Rush Hours:** Increase the number of buses and trains strictly during peak commute times (**7-9 AM** and **4-7 PM**) to give drivers an easy alternative.
4. **Rush Hour Truck Bans:** Limit large delivery trucks from using major roads during peak hours to keep lanes open for commuters.
5. **Peak-Time Toll Adjustments:** Charge slightly higher tolls during peak hours to encourage non-essential drivers to travel at off-peak times.

3. **Speed Harmonization Measures:**
   * Utilize variable speed limits during high-volume periods to smooth out traffic flow, reduce stop-and-go driving, and delay the onset of severe congestion.

* **Congestion Friction & Low Speeds:** Accident frequency increases sharply in high-congestion, low-speed environments. Stop-and-go driving conditions elevate the probability of minor and secondary rear-end collisions.
* **Secondary Incident Cascades:** Active roadway incidents are the single strongest precursor to secondary accidents. The presence of existing obstacles drastically increases the likelihood of additional collisions on that same corridor.

### 🎯 Strategic Recommendations

   * Implement automated variable speed displays and dynamic warning signs during severe weather events to reduce vehicle approach speeds before high-risk zones.

2. **Rapid Incident Clearance Protocol:**
   * Establish priority towing and quick-clearance response teams along high-traffic corridors to remove initial road obstacles rapidly and prevent secondary collision chains.

3. **Congestion-Aware Emergency Services:**
   * Pre-position emergency response units near historical bottleneck zones during peak hours to shorten incident clearance times and maintain lane availability.

# 11. How do major city events affect mobility and public transport operations?

```python
City_Events['start_timestamp'] = pd.to_datetime(City_Events['start_timestamp'])
City_Events['end_timestamp'] = pd.to_datetime(City_Events['end_timestamp'])
Traffic['timestamp'] = pd.to_datetime(Traffic['timestamp'])
Public_Transport['timestamp'] = pd.to_datetime(Public_Transport['timestamp'])

# Merge mobility datasets
mobility = pd.merge(Traffic, Public_Transport, on=['timestamp', 'district_id'], how='inner')

# Flag event windows
mobility['event_period'] = 'Normal'

for _, event in City_Events.iterrows():
  dist = event['district_id']
  start = event['start_timestamp']
  end = event['end_timestamp']

  d_mask = mobility['district_id'] == dist

  # During Event
  mobility.loc[
      d_mask & (mobility['timestamp'] >= start) & (mobility['timestamp'] <= end),
      'event_period',
  ] = 'During Event'

  # Before Event (3h Window)
  mobility.loc[
      d_mask
      & (mobility['timestamp'] >= start - pd.Timedelta(hours=3))
      & (mobility['timestamp'] < start)
      & (mobility['event_period'] != 'During Event'),
      'event_period',
  ] = 'Before Event'

  # After Event (3h Window)
  mobility.loc[
      d_mask
      & (mobility['timestamp'] > end)
      & (mobility['timestamp'] <= end + pd.Timedelta(hours=3))
      & (mobility['event_period'] != 'During Event'),
      'event_period',
  ] = 'After Event'

# Aggregate metrics across phases
event_impact = (
    mobility.groupby('event_period')
    .agg(
        avg_traffic_volume=('traffic_volume', 'mean'),
        avg_passengers=('passenger_count', 'mean'),
        avg_speed=('average_speed_kmh', 'mean'),
        avg_pt_delay=('average_delay_minutes', 'mean'),
    )
    .reindex(['Normal', 'Before Event', 'During Event', 'After Event'])
    .reset_index()
)

# Plot Setup
plt.style.use('seaborn-v0_8-whitegrid')
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
plt.subplots_adjust(hspace=0.35, wspace=0.25)

# 1. Traffic Volume Across Phases
sns.barplot(
    data=event_impact,
    x='event_period',
    y='avg_traffic_volume',
    palette='Blues',
    ax=axes[0, 0],
)
axes[0, 0].set_title(
    'Average Traffic Volume Across Event Phases', fontsize=12, fontweight='bold'
)
axes[0, 0].set_xlabel('', fontsize=10)
axes[0, 0].set_ylabel('Traffic Volume (vehicles/hr)', fontsize=10)

# 2. Public Transport Ridership Across Phases
sns.barplot(
    data=event_impact,
    x='event_period',
    y='avg_passengers',
    palette='Purples',
    ax=axes[0, 1],
)
axes[0, 1].set_title(
    'Public Transport Ridership Across Event Phases',
    fontsize=12,
    fontweight='bold',
)
axes[0, 1].set_xlabel('', fontsize=10)
axes[0, 1].set_ylabel('Passenger Count', fontsize=10)

# 3. Vehicle Speed Across Phases
sns.barplot(
    data=event_impact,
    x='event_period',
    y='avg_speed',
    palette='Reds_r',
    ax=axes[1, 0],
)
axes[1, 0].set_title(
    'Average Vehicle Speed (km/h) Across Event Phases',
    fontsize=12,
    fontweight='bold',
)
axes[1, 0].set_xlabel('', fontsize=10)
axes[1, 0].set_ylabel('Speed (km/h)', fontsize=10)

# 4. Public Transport Delay Across Phases
sns.barplot(
    data=event_impact,
    x='event_period',
    y='avg_pt_delay',
    palette='Oranges',
    ax=axes[1, 1],
)
axes[1, 1].set_title(
    'Average Public Transport Delay Across Event Phases',
    fontsize=12,
    fontweight='bold',
)
axes[1, 1].set_xlabel('', fontsize=10)
axes[1, 1].set_ylabel('Average Delay (Minutes)', fontsize=10)

plt.tight_layout()
plt.show()
```

## 💡 Insights & Strategic Recommendations

### 📊 Key Insights

* **Massive Volume & Demand Surge During Events:** Traffic volume more than doubles during active city events compared to normal baseline levels, while public transport passenger counts jump significantly as attendees flock to venue districts.
* **Severe Speed Collapse & Congestion Spike:** Vehicle speeds drop rapidly during event windows as road capacity saturates, raising congestion scores significantly above normal baseline levels.
* **Transit Delay Cascades:** High road congestion during major events spills over into public transport operations, causing average bus and transit delays to rise during active event windows.
* **Rapid Post-Event Dispersal:** Dispersal following event conclusion occurs efficiently, with traffic volume dropping below normal baseline levels and vehicle speeds quickly recovering.

---

### 🎯 Strategic Recommendations

1. **Pre-Event Dynamic Transit Fleet Scaling:**
   * Automatically deploy extra shuttle buses and increase rail frequency to host districts during the three-hour buildup window prior to event start times to absorb incoming demand.

2. **Event-Specific Traffic Control & Lane Prioritization:**
   * Establish temporary bus-only lanes and dynamic traffic signal green-waves leading to major venues to protect public transit speed and prevent delay cascades during active event hours.

3. **Post-Event Rapid Evacuation Routes:**
   * Designate dedicated egress corridors for high-capacity public transport immediately following event end times to maintain rapid crowd dispersal.

```python
# Format hourly timestamps across pre-loaded DataFrames
Emergency_Events['timestamp_hour'] = pd.to_datetime(
    Emergency_Events['timestamp']
).dt.floor('h')
Traffic['timestamp_hour'] = pd.to_datetime(Traffic['timestamp']).dt.floor('h')
Weather['timestamp_hour'] = pd.to_datetime(Weather['timestamp']).dt.floor('h')

# Merge Emergency_Events, Traffic, Weather, and Districts
Em_DF = Emergency_Events.merge(
    Traffic, on=['timestamp_hour', 'district_id'], how='inner'
)
Em_DF = Em_DF.merge(
    Weather, on=['timestamp_hour', 'district_id'], how='inner'
)
Em_DF = Em_DF.merge(Districts, on='district_id', how='inner')

# Setup $2 \times 2$ Grid Plot
plt.style.use('seaborn-v0_8-whitegrid')
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
plt.subplots_adjust(hspace=0.35, wspace=0.25)

Sample_EM = Em_DF.sample(n=min(5000, len(Em_DF)), random_state=42)

# 1. Congestion Index vs Response Time
sns.regplot(
    data=Sample_EM,
    x='congestion_index',
    y='response_time_minutes',
    scatter_kws={'alpha': 0.15, 'color': '#dc2626'},
    line_kws={'color': 'black'},
    ax=axes[0, 0],
)
axes[0, 0].set_title(
    'Congestion Index vs. Response Time', fontsize=12, fontweight='bold'
)
axes[0, 0].set_xlabel('Congestion Index', fontsize=10)
axes[0, 0].set_ylabel('Response Time (Minutes)', fontsize=10)

# 2. Average Speed vs Response Time
sns.regplot(
    data=Sample_EM,
    x='average_speed_kmh',
    y='response_time_minutes',
    scatter_kws={'alpha': 0.15, 'color': '#2563eb'},
    line_kws={'color': 'red'},
    ax=axes[0, 1],
)
axes[0, 1].set_title(
    'Average Speed (km/h) vs. Response Time', fontsize=12, fontweight='bold'
)
axes[0, 1].set_xlabel('Average Speed (km/h)', fontsize=10)
axes[0, 1].set_ylabel('Response Time (Minutes)', fontsize=10)

# 3. Weather Condition vs Response Time
Weather_Resp = (
    Em_DF.groupby('weather_condition')['response_time_minutes']
    .mean()
    .reset_index()
    .sort_values(by='response_time_minutes', ascending=False)
)
sns.barplot(
    data=Weather_Resp,
    x='weather_condition',
    y='response_time_minutes',
    palette='Reds_r',
    ax=axes[1, 0],
)
axes[1, 0].set_title(
    'Average Response Time by Weather Condition', fontsize=12, fontweight='bold'
)
axes[1, 0].set_xlabel('', fontsize=10)
axes[1, 0].set_ylabel('Mean Response Time (Minutes)', fontsize=10)
axes[1, 0].tick_params(axis='x', rotation=20)

# 4. Severity Level vs Response Time
Sev_Order = ['Low', 'Medium', 'High', 'Critical']
Sev_Resp = (
    Em_DF.groupby('severity_level')['response_time_minutes']
    .mean()
    .reindex([s for s in Sev_Order if s in Em_DF['severity_level'].unique()])
    .reset_index()
)
sns.barplot(
    data=Sev_Resp,
    x='severity_level',
    y='response_time_minutes',
    palette='Oranges_r',
    ax=axes[1, 1],
)
axes[1, 1].set_title(
    'Average Response Time by Emergency Severity Level',
    fontsize=12,
    fontweight='bold',
)
axes[1, 1].set_xlabel('Severity Level', fontsize=10)
axes[1, 1].set_ylabel('Mean Response Time (Minutes)', fontsize=10)

plt.tight_layout()
plt.show()
```

# 13. Which districts experience the highest emergency burden, and what factors characterize them?

```python
# Format hourly timestamps across pre-loaded DataFrames
Emergency_Events['timestamp_hour'] = pd.to_datetime(
    Emergency_Events['timestamp']
).dt.floor('h')
Traffic['timestamp_hour'] = pd.to_datetime(Traffic['timestamp']).dt.floor('h')
Weather['timestamp_hour'] = pd.to_datetime(Weather['timestamp']).dt.floor('h')

# Merge DataFrames
Em_District_DF = Emergency_Events.merge(
    Districts, on='district_id', how='inner'
)
Em_District_DF = Em_District_DF.merge(
    Traffic, on=['timestamp_hour', 'district_id'], how='inner'
)
Em_District_DF = Em_District_DF.merge(
    Weather, on=['timestamp_hour', 'district_id'], how='inner'
)

# Calculate District Level Metrics
District_Burden = (
    Em_District_DF.groupby(['district_id', 'district_name', 'district_type'])
    .agg(
        total_emergencies=('event_id', 'count'),
        avg_response_time=('response_time_minutes', 'mean'),
        ambulance_req_pct=('ambulance_required', lambda x: x.mean() * 100),
        hospital_transport_pct=(
            'hospital_transport_required',
            lambda x: x.mean() * 100,
        ),
        population_density=('population_density', 'first'),
        avg_congestion=('congestion_index', 'mean'),
    )
    .reset_index()
    .sort_values(by='total_emergencies', ascending=False)
)

plt.style.use('seaborn-v0_8-whitegrid')
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
plt.subplots_adjust(hspace=0.35, wspace=0.25)

Top_Districts = District_Burden.head(8)

# 1. Total Emergencies by District
sns.barplot(
    data=Top_Districts,
    x='district_id',
    y='total_emergencies',
    palette='Reds_r',
    ax=axes[0, 0],
)
axes[0, 0].set_title(
    'Top Districts by Total Emergency Incidents', fontsize=12, fontweight='bold'
)
axes[0, 0].set_xlabel('District ID', fontsize=10)
axes[0, 0].set_ylabel('Total Emergency Events', fontsize=10)

for p in axes[0, 0].patches:
  axes[0, 0].annotate(
      f'{int(p.get_height())}',
      (p.get_x() + p.get_width() / 2.0, p.get_height()),
      ha='center',
      va='center',
      xytext=(0, 5),
      textcoords='offset points',
      fontsize=9,
      fontweight='bold',
  )

# 2. Emergency Type Breakdown in Top Districts
Top_District_IDs = Top_Districts['district_id'].tolist()
Filtered_Em = Em_District_DF[
    Em_District_DF['district_id'].isin(Top_District_IDs)
]
sns.countplot(
    data=Filtered_Em,
    x='district_id',
    hue='emergency_type',
    palette='Set2',
    ax=axes[0, 1],
)
axes[0, 1].set_title(
    'Emergency Type Breakdown in High-Burden Districts',
    fontsize=12,
    fontweight='bold',
)
axes[0, 1].set_xlabel('District ID', fontsize=10)
axes[0, 1].set_ylabel('Incident Count', fontsize=10)
axes[0, 1].legend(title='Emergency Type', fontsize=9)

# 3. Medical Resource & Transport Requirements
Trans_DF = Top_Districts.melt(
    id_vars=['district_id'],
    value_vars=['ambulance_req_pct', 'hospital_transport_pct'],
    var_name='Requirement_Type',
    value_name='Percentage',
)
Trans_DF['Requirement_Type'] = Trans_DF['Requirement_Type'].map({
    'ambulance_req_pct': 'Ambulance Required (%)',
    'hospital_transport_pct': 'Hospital Transport (%)',
})

sns.barplot(
    data=Trans_DF,
    x='district_id',
    y='Percentage',
    hue='Requirement_Type',
    palette='Oranges_r',
    ax=axes[1, 0],
)
axes[1, 0].set_title(
    'Medical Resource & Transport Requirements by District',
    fontsize=12,
    fontweight='bold',
)
axes[1, 0].set_xlabel('District ID', fontsize=10)
axes[1, 0].set_ylabel('Requirement Rate (%)', fontsize=10)
axes[1, 0].legend(title='', fontsize=9)

# 4. Population Density vs Total Emergencies
sns.scatterplot(
    data=District_Burden,
    x='population_density',
    y='total_emergencies',
    size='avg_congestion',
    hue='avg_congestion',
    sizes=(50, 300),
    palette='Reds',
    ax=axes[1, 1],
)
for _, row in District_Burden.iterrows():
  axes[1, 1].annotate(
      row['district_id'],
      (row['population_density'], row['total_emergencies']),
      textcoords='offset points',
      xytext=(0, 5),
      ha='center',
      fontsize=8,
      fontweight='bold',
  )
axes[1, 1].set_title(
    'Population Density vs. Total Emergency Burden',
    fontsize=12,
    fontweight='bold',
)
axes[1, 1].set_xlabel('Population Density (people/km²)', fontsize=10)
axes[1, 1].set_ylabel('Total Emergencies', fontsize=10)

plt.tight_layout()
plt.show()
```

# 14. Which districts face the highest combined urban operational pressure?

```python
# Format hourly timestamps across pre-loaded DataFrames
Emergency_Events['timestamp_hour'] = pd.to_datetime(
    Emergency_Events['timestamp']
).dt.floor('h')
Traffic['timestamp_hour'] = pd.to_datetime(Traffic['timestamp']).dt.floor('h')
Weather['timestamp_hour'] = pd.to_datetime(Weather['timestamp']).dt.floor('h')
Air_Quality['timestamp_hour'] = pd.to_datetime(
    Air_Quality['timestamp']
).dt.floor('h')
Power_Grid['timestamp_hour'] = pd.to_datetime(
    Power_Grid['timestamp']
).dt.floor('h')
Public_Transport['timestamp_hour'] = pd.to_datetime(
    Public_Transport['timestamp']
).dt.floor('h')

# 1. District Level Aggregations
mob = Traffic.groupby('district_id')['congestion_index'].mean().reset_index()
env = Air_Quality.groupby('district_id')['synthetic_aqi'].mean().reset_index()
eng = Power_Grid.groupby('district_id')['grid_stress_index'].mean().reset_index()
emerg = (
    Emergency_Events.groupby('district_id')['event_id']
    .count()
    .reset_index()
    .rename(columns={'event_id': 'emergency_count'})
)
pt_press = (
    Public_Transport.groupby('district_id')['average_delay_minutes']
    .mean()
    .reset_index()
)

# Merge dimensions into Districts DataFrame
df_score = Districts[
    ['district_id', 'district_name', 'district_type']
].merge(mob, on='district_id')
df_score = df_score.merge(env, on='district_id')
df_score = df_score.merge(eng, on='district_id')
df_score = df_score.merge(emerg, on='district_id')
df_score = df_score.merge(pt_press, on='district_id')


# Min-Max Normalization (0 to 100)
def min_max_scale(series):
  return (series - series.min()) / (series.max() - series.min()) * 100


df_score['Mobility_Score'] = min_max_scale(df_score['congestion_index'])
df_score['Environmental_Score'] = min_max_scale(df_score['synthetic_aqi'])
df_score['Energy_Score'] = min_max_scale(df_score['grid_stress_index'])
df_score['Emergency_Score'] = min_max_scale(df_score['emergency_count'])
df_score['Public_Transport_Score'] = min_max_scale(
    df_score['average_delay_minutes']
)

# Calculate Composite Score
score_cols = [
    'Mobility_Score',
    'Environmental_Score',
    'Energy_Score',
    'Emergency_Score',
    'Public_Transport_Score',
]
df_score['Composite_Score'] = df_score[score_cols].mean(axis=1)
df_score = df_score.sort_values(by='Composite_Score', ascending=False)

# Setup 1x2 Synthesis Plot
plt.style.use('seaborn-v0_8-whitegrid')
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# Plot 1: Top 10 Composite Ranking
top10_score = df_score.head(10)
sns.barplot(
    data=top10_score,
    x='district_id',
    y='Composite_Score',
    palette='Reds_r',
    ax=axes[0],
)
axes[0].set_title(
    'Top 10 Districts by Composite Urban Operational Pressure',
    fontsize=13,
    fontweight='bold',
)
axes[0].set_xlabel('District ID', fontsize=11)
axes[0].set_ylabel('Composite Pressure Score (0-100)', fontsize=11)

for p in axes[0].patches:
  axes[0].annotate(
      f'{p.get_height():.1f}',
      (p.get_x() + p.get_width() / 2.0, p.get_height()),
      ha='center',
      va='center',
      xytext=(0, 5),
      textcoords='offset points',
      fontsize=9,
      fontweight='bold',
  )

# Plot 2: Heatmap Dimension Breakdown
heatmap_data = top10_score.set_index('district_id')[score_cols]
heatmap_data.columns = [
    'Mobility',
    'Environmental',
    'Energy',
    'Emergency',
    'Transit',
]

sns.heatmap(
    heatmap_data,
    cmap='YlOrRd',
    annot=True,
    fmt='.1f',
    cbar_kws={'label': 'Normalized Pressure Score (0-100)'},
    ax=axes[1],
)
axes[1].set_title(
    'Urban Operational Pressure Breakdown (Top 10 Districts)',
    fontsize=13,
    fontweight='bold',
)
axes[1].set_xlabel('Operational Dimension', fontsize=11)
axes[1].set_ylabel('District ID', fontsize=11)

plt.tight_layout()
plt.show()
```
